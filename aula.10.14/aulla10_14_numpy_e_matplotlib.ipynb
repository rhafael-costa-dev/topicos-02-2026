```python
# Import das bibliotecas
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Criação da lista e do numpy array
dados = [100, 120, 150, 90, 180]
vendas = np.array([100, 120, 150, 90, 180])
```

```python
# Calculo da média com dados da lista em python
soma = 0
for v in dados:
  soma += v

media = soma / len(dados)
print(media)
```

```python
print(vendas) # print do array
print(type(vendas)) # print do tipo do objeto array
print(vendas.mean()) # print da média dos dados do array
```

```python
zeros = np.zeros((2, 3), dtype=float)
print(zeros.ndim)
print(zeros.shape)
print(zeros.dtype)
print(zeros)
```

```python
bonus = vendas - 10
print(vendas)
print(bonus)
```

```python
notas = np.array([7.0, 8.5, 10.0, 6.0, 9.0])

print(f" Menor: {notas.min():.2f}")
print(f" Maior: {notas.max():.2f}")
print(f" Media: {notas.mean():.2f}")
print(f"  Soma: {notas.sum():.2f}")
print(f"Desvio: {notas.std():.2f}")
```

```python
print(len(notas))
print(notas[4])
print(notas[2:])
print(notas[:3])
print(notas[4:10])
```

```python
dias = [1, 2, 3, 4, 5]
vendas = [100, 120, 150, 90, 180]

plt.plot(dias, vendas)
plt.show()
```

```python
produtos = ['Pizza', 'Hambúrguer', 'Refrigerante']
vendas = [100, 80, 150]

plt.bar(produtos, vendas, color='#2563eb')
plt.title('Vendas por Produto')
plt.ylabel('Unidades Vendidas')
plt.xlabel('Produtos')
plt.show()
```

```python
sabores = ['Calabresa', 'Mussarela', 'Frango']
vendas = [50, 25, 25]
plt.pie(vendas, labels=sabores, autopct='%1.2f%%')
plt.title('Distribuição de Sabores Vendidos')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(8, 3))
# Gráfico 1 (Linha na esquerda)
ax[0].plot([1, 2, 3], [10, 20, 15])
ax[0].set_title('Evolução')

# Gráfico 2 (Barras na direita)
ax[1].bar(['A', 'B'], [5, 12])
ax[1].set_title('Comparação')

plt.show()
```
